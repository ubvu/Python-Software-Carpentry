# Analyzing Data from Multiple Files

### Listing file paths
When one needs to apply the same task on different files, it can be usefull to be able to list those files without having to write their names out one by one. Let's checkout how this can be done with glob.

```python
import glob # import glob
```

```python
glob.glob("data/*.csv") # glob.glob list all the files with the structure that is given. 
                        # In this case, all csv files in the data folder, as the * is a wildcard, as they follow 
                        # that pattern. Where there can be any number of characters in the place of the asterisk, 
                        # the question mark can be used for precisely one character.

```

```python
# not all computers will list the files in the same order, when using glob.glob(). The using the sorted() function
# will make sure the files are sorted in the same way. Using this will help with reproducibility. 
sorted(glob.glob("data/*.csv"))
```

```python
# Here follows a way to extract the file name from the path
glob.glob("data/*.csv")[0].split("\\")[-1]
```

```python
# print the filenames one by one making use of a for loop
files = glob.glob("data/*.csv")

for file in files:
    print(file)
```

    # import the relevant packages, if you haven't done so already
    import pandas as pd 
    import glob
    
    
    filenames = glob.glob("data/gapminder_gdp_*.csv") # list the files from the different continents
    countries = [] # initialize a list that we can append the countries in
    
    for file in filenames: # loop over the files
        df = pd.read_csv(file, index_col = 'country') # open the file
        countries = countries + list(df.index) # create a list of the countries (which are a part of the index), and add them to the list
    
    print(len(countries), 'have a record in the gdp data')
