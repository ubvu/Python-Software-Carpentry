```python
# We will use a new library here called glob (not sucha an appealing name, but very useful)

import glob
```

```python
# glob has a cool capability to list all your files (and folders) inside a directory
# First we will use the "*" known as a wildcard, it allows anything to fill this position, any character and any number of them
# This code will go into the data folder and show all csv files that have "gapminder" at the start of its filename

glob.glob("data/gapminder*.csv")
```

```python
# if we wanted to see all the files in the folder we can just use the *
# in this situation the data folder should only have files that are gapminder ones
# (I accidentally saved a notebook in there too)

glob.glob("data/*")
```

```python
# Take a look at the output, it is returned as a list
# for some systems this won't be in alphabetical order (or alphanumeric order)
# to order the list (or any list) we use the sorted function

sorted(glob.glob("data/*"))
```

```python
# Where the * replaces any number of characters (letters, numbers or special characters like "&")
# the ? is a placeholder for a single character
# Let's test it out

sorted(glob.glob("data/gapminder_gdp_??????.csv"))
# We get two results as they both have 6 characters following "gapminder_gdp_"
```

```python
# Getting a list back is very useful, because we can then index the list
# We could even loop over the list
# Very useful for when you want to repeat an analysis or cleaning over many datasets

filenames = sorted(glob.glob("data/gapminder*.csv"))
```

```python
for filename in filenames:
    print(filename)

# Note that glob will give the full relative path from where you are running the code
# So even though we ask it wo look inside the data folder all paths still include the data folder 
```

```python
# Can you print the number of countries recorded per region?
import pandas as pd
import glob
filenames = glob.glob("data/*.csv")
print(filenames)
countries = []
for file in filenames:
    df = pd.read_csv(file)
    countries += list(df["country"].unique())
print(len(countries))
```
