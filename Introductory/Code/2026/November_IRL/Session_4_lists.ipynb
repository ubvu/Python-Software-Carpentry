There are also risks of updating mutable objects in place. Take a look at the next example where we create a list of ingredients for a recipe and use it to make another and exchange one ingredient for another.

```python
mild_salsa = ["peppers", "onions", "coriander", "tomatoes"]
hot_salsa = mild_salsa
hot_salsa[0] = "scotch bonnet" # Exchanging the peppers for the far hotter scotch bonnets
# printing both to check out our ingredient lists
print("ïngredients mild salsa: ", mild_salsa)
print("ingredients hot salsa: ", hot_salsa)
```

Well that didn't work out. The issue is that when we created hot_salsa we actually just provided Python with a second name to use for the list. Under the hood Python has stored the list in a place in the computers memory and the name mild salsa points to that. When writing ``hot_salsa = mild_salsa`` another name began to point to it.

```python
# To fix this we can apply the list constructor to mild salsa and it will build a new list from the reference of mild_salsa
mild_salsa = ["peppers", "onions", "coriander", "tomatoes"]
hot_salsa = list(mild_salsa)
hot_salsa[0] = "carolina reaper" # A New hotter pepper because I don't like my guests
print("mild", mild_salsa)
print("hot", hot_salsa)
```

Above we used a constructor (special function) to create a new list from the reference. Constructors can be found for all data types in python and will convert the object to the new type (where possible).

```python
# Slicing
# SLicing is the term for using a range in the indexing of an indexable object
# Slicing is done with a ':' and the number of the left is the position where your selection begins 
# and the number on the right is the first position nnot included
binomial_name = "Drosophila melanogaster"
group = binomial_name[0:10]
print("group: ", group)
```

```python
# We don't have to start at zero!
species = binomial_name[11:23]
print(species)
```

```python
# Another inbuilt function is `len()`, it returns the length of an object
# in this case it checks the number of characters in binomial name
len(binomial_name)
```

```python
# Using the length of the list to get the full range
print("the species is ", binomial_name[11:len(binomial_name)])
```

```python
# You can use a third value in your slicing to set the step size. This detemines how many elements the section moves
# along by. The default size is one. 
test_list = [1,2,3,4,5,6,6,76,87,8,10]
print(test_list[1:9:3])
# Can you use the step size argument to create a new string that contains only every other character and store it in a new variable
beatles = "In an octopus's garden in the shade"
```

```python
print(beatles[ : :2]) # As we saw with test list the third value is the step
# This lets the default value fill in for the start and end of the list and the third value makes it skip every other character
```
