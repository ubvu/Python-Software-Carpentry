In this part we will make a plot. To do that we make use of the matplotlib library, specifically the pyplot module.

```python
# Due to convention (and its long name) we will provide matplotlib.pyplot with an alias, plt
import matplotlib.pyplot as plt
```

```python
# First we make some data to plot
time = [0, 1, 2, 3]
position = [0, 100, 200, 300]
# This function will give us a simple line graph
plt.plot(time, position)
plt.xlabel("time (hr)") # Adding labels for the x and y axis
plt.ylabel("position (km)")
```
