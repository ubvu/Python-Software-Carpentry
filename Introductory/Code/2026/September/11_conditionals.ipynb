```python
# We start by defining a value for a variable and using an if statement
# The if statement checks the condition in the statement and if it is True then it will do what is inside the statement
# Like with loops we use ":" to indicate the end of the statement
# Everything you want python to do when the condition is met should be indented (with a tab)

number = 101

if number > 100:
    print("number is greater than 100")

if number <= 100:
    print("number is equal to or smaller than 100")
    
```

```python
# Mathematical functions

# > : greateer than
# < : less than
# == : equal to (different from "=")
# != : not equal to
# >= : greater than or equal to
# <= : less than or equal to
```

```python
# We can also use the else statement
# It means in every other situation, do the following

number = 101

if number > 100:
        print("number is greater than 100")
else:
    print("number is equal to or smaller than 100")
```

```python
# There is a third statement kind, that is else if or elif
# That means within the block of statements if condition 1 is met do the following, else if condition 2 is met do the following
# if we include another if or an else it is the end of our conditional block

if number > 100:
        print("number is greater than 100")
elif number == 100:
    print("the number is 100!")
else:
    print("number is smaller than 100")
```

```python
# We can use more than one elif
# So let's give it a go

number = -1

if number > 100:
        print("number is greater than 100")
elif number == 100:
    print("the number is 100!")
elif number < 0 :
    print("error this number is negative")
else:
    print("number is smaller than 100")
```

```python
# This can also be done with strings
# And we can also do multiple checks in one condition by using "or"

banana = "purple"

if banana == "green" or "yellow":
    print("eat")
elif banana == "brown":
    print("make it into banana bread")
else:
    print("buy new bananas")
```

```python
# hmm that didn't work as expected
# Well that is because we didn't ask python to check that the variable banana == "yellow"
# We actually just said if banana is equal to "green" or the string "yellow" exists
# Let's test that out by just having a string

if "blue":
    print("colourful")
```

```python
# a clearer example might be setting a variable directly as True or False
# here we look at birthdays

is_it_my_birthday = False

if is_it_my_birthday:
    print("Let's eat cake!")
```

```python
# Let's fix the banana code then with our new knowledge

banana = "purple"

if banana == "green" or banana == "yellow":
    print("eat")
elif banana == "brown":
    print("make it into banana bread")
else:
    print("buy new bananas")
```

```python
# Let's add more complexity to our banana example
# In the situation that we have dots on our banana we think it is ready for a smoothie
# without dots it is ready for the bin

banana = "brown"
dots = True

if banana == "green" or banana == "yellow":
    print("eat")
elif banana == "brown" and dots:
    print("make it into a smoothie")
else:
    print("throw it in the bin :( ")
```
