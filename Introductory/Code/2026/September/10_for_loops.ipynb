```python
# Let's start with a small test data set in the form of a list of odd numbers

odds = [1,3,5]

# We are going to use a for loop
# Here we are asking python to loop over the elements of odds
# It creates the variable num that will hold the value of the 1st element of odds, then the second then the third
# The execution of the loop results in it printing the value of num
# Here we have used num as our variable name, but we could have chosen anything
# num is not a specific built in name or command in python, these are highlighted green
for num in odds:
    print(num)
```

```python
# We can also use range, a function that 'contains' all the values from 0 to 3
# If we want to change the starting point of range we can, e.g. range(2,5) will be 2,3,4
# range goes up to, but not including, the number in the brackets

for num in range(3):
    print(num)
```
