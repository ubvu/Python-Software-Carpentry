The great thing about programming is that people share their work. Functions perform specific acts and a script that has a several functions in is called a package. When you have multiple scripts tied together to provide different functions this is called a library (the scripts are called modules). 

We can make use of these libraries to speed up our own work instead of writing the code on our own. There are some libraries that come with Python, such as the math library.

To use these libraries we first need to import them. This tells python, "hey, I want to make use of the functions here. Load up the scripts in the background"

```python
# Import is a key word, so it is highlighted green
import math
```

```python
# We can make use of a value that the math library contains, pi
# To access this we use the ".", this says to python 
# "look at the math library I told you to import, inside of that is something called pi, I want that"
print("pi is:", math.pi)
```

```python
# when using the wrong name we get a console print out
import maths
```

```python
# math also has functions, here we use the cos function on the pi value
# functions require that you use the () after their name
print("cos(pi) is: ", math.cos(math.pi))
```

```python
# We can always use the inbuilt help() function to get more information about the library or its functions
help(math)
```

```python
# If we are often using cos and pi we might not want to type out "math.cos" each time
# This means that we can use the name cos or pi directly instead
from math import cos, pi

print("cos(pi) is:", cos(pi))
```

```python
# Python also allows for aliases of libraries
# When a library has a long name this can save a lot of time and make your code more readable
import math as m

print("cos(pi)", m.cos(m.pi))
```
