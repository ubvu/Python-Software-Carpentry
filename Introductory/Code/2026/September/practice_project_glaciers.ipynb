### A brief introduction into the glacier projections
The Open Global Glacier Model ([OGGM](https://oggm.org/)) is an open source model. It's one of the models that is being used to make projections of glacier mass loss as a result of a changing climate over the 21<sup>st</sup> century. For the <i>OGGM standard projections</i>, the model has been forced with data from climate model simulations, from multiple models and for a range of scenarios. Here we will analyse these glacier simulations ourselves. 

There are about ~200.000 glacier globally. Though these glaciers have been simulated individually, we will use the data that has been compiled either by region or globally. To keep things simple for this tutorial, we will not worry about glaciers that are missing from the simulations.

The unit m<sup>3</sup> is a little awkward for analysing and visualizing glacier evolution at a region to global level. The next step is to convert to more proportional unit like km<sup>3</sup>, Gt or mm sea level rise equivalent (SLE). 

volume (km<sup>3</sup>) = volume (m<sup>3</sup>) / 10<sup>9</sup>

mass of ice (Gt) = volume (km<sup>3</sup>) * 0.9

SLE (mm) = mass of ice (Gt) x (1 / 361.8)
