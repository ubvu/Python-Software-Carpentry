```python
# our first variable
# We can store values in variables to refer to them by name
# this is done with a "="
weight_kg = 60
```

Most IDEs (Interactive Development Environments, places to write and run code) have colour coding based on the language to help make it more readable. In this notebook operators are purple, comments are a sort of gray blue, numbers are green, and variables are black. 

```python
# This time python didn't show the value, we have to ask it to show it to us
weight_kg
```

```python
# Notebooks will show you the value of a variable if it is on the last line of a cell
# Python scripts don't do this and so for compatibility and clarity we use the print function
# functions require round brackets () and the 'argument(s)' goes inside them
# Inbuilt functions are also highlighted green
print(weight_kg)
```

```python
# look at the output here, there is one 60 and one 25
# the second weight_kg is not shown because it is not the last line of the cell
print(weight_kg)
weight_kg
5 * 5
```

```python
# I spent weeks eating well, resting up and working out and made some progress!
# updating our variable
weight_kg = 60.3
```

```python
print(weight_kg) # Checking the update worked
```

```python
# Variables can also hold other data types
# Text is referred to as a string type
# strings can be written with either "" or ''
location = "uni gym"
print(location)
```

```python
# I have a cousin in the US who also goes to the gym, in order to let him know about my progress I have to convert the units to pounds (lbs)
# Thankfully we can create a variable using another variable
weight_lb = 2.2 * weight_kg
print(weight_lb)
```

```python
# The same works for strings
lift_info = "October, " + location
```

```python
print(weight_lb)
print(lift_info)
```

```python
# using more than one argument in the print
print(weight_lb, lift_info)
```

```python
# You are also able to use raw data as input to the print function, it doesn't need to be a variable
print("weight in pounds", weight_lb, lift_info)
```

```python
# There is another inbuilt function, type() that checks the data type of an object
# The print function can even have another function call inside it
# Python will first evaluate type(60) and type(60.3) before calling print
print(type(60), type(60.3))
```

```python
# Variables can be used in the type function too!
print(type(lift_info))
```
