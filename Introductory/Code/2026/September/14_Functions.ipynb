```python
# Our in class solution
def fahr_to_kelvin(temp_f):
    return ((temp_f - 32) * (5/9)) + 273.15
print("the freezing temperature of water is: ", fahr_to_kelvin(32), "K")
```

```python
# We also have the option of using the functions that we have already written
# This is a sort of pipeline with the output of one function feeding into another
# This is what we do in larger projects, each function has a small role to play in the bigger story
def fahr_to_kelvin(temp_f):
    temp_c = fahr_to_celsius(temp_f)
    temp_k = celsius_to_kelvin(temp_c)
    return temp_k
```

```python
# Testing the function
print("boiling temperature of water in kelvin is: ", fahr_to_kelvin(212))
```

```python
# We saw before that you can add (or concatenate) strings, can you write a function that will 
# “wrap” a string inside another string such that
# “print(fence(‘name’, ‘*’))”
# Will give the output
# “*name*”

def fence(string_1, wrapper):
    string_2 = wrapper + string_1 + wrapper
    return string_2
# testing our function
fence("henry", "~~")
```

```python
# There are two arguments needed, so both must exist in the brackets of the function definition
# We can also mulitply the wrapper as we wish
def fence(string_1, wrapper):
    string_2 = (wrapper * 4)+ string_1 + (wrapper * 2)
    return string_2

fence("mike", "o")
```

```python
# This time we have a default value for wrapper
# By including an "=" and a value after wrapper we ahve set a default value for it
# We know that for most of the uses of this function we want to wrap with 3 *'s so we set it as the default
# Many functions you will use will have default values, often for majority of the arguments

def fence(string_1, wrapper="*"):
    string_2 = (wrapper * 3) + string_1 + (wrapper * 3)
    return string_2
# we can run this without the second argument
fence("ben")
```

```python
# We can still provide our own wrapper if we want and the function will use the one we supply
fence("pam", "$")
```

```python
# We can provide documentation for our own functions
# That means that we (or other people) can use the help function on our own functions to remember (or find out) what it does
# This can be done by adding what is known as a doc-string
# We place some text describing the function between a set of triple double quotes """
# Typically people also list the arguments and their types and describe the output of the function

def my_func(name, colour):
    """
    This function addresses the user and asks about their favourite colour.
    
    args: 
    name - a string that is the name of someone or something (living)
    colour - a colour that might be your favourite
    output: 
    a message to the user addressing them and asking their favourite colour
    """
    msg = "hello " + name + "! is your favourite colour " + colour + "?"
    return msg

my_func("bluebell", "silver")
```
