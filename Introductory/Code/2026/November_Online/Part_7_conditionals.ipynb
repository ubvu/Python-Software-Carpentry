# Making Choices

It can be usefull to be able to apply an action only in a certain case. This notebook is about coding conditions.

````python
number = 100 # We will use the variable number here as an example.
````

````python
number > 100 # With ">" we can check it the value the variable number has attached great than the another value (100 in this case).
             # The result of this will be boolean (either True or False).
````

It's not only possible to check if something is larger than. Here is a list of the mathematical functions that can be used in a comparison:

 ```>``` : greater than
 
 ```<``` : smaller than
 
 ```==``` : equal to
 
 ```!=``` : not equal to
 
 ```>=``` : greater than or equal to
 
 ```<=``` : smaller than or equal to

```True``` and ```False``` can be used to build conditional statements, or in other words if-statements.

An if-statement has some similarity to the for loop. 
- Instead of ```for``` it only always starts with ```if```.
- Followed by the condition that ends with ```:```.
- The body of the statement needs to be intended.

````python
if number > 100: 
    print("number is greater than 100.")
````

Nothing happens when the condition is ```False```. More if-statements can be added to explore the different possibilities.

````python
if number > 100: 
    print("number is greater than 100.")

if number == 100: 
    print("the number is 100")
    
if number >= 100:
    print("number is greater than or equal too 100.")
````

However there is a more practical way to deal with this. It is possible to check multiple conditions in a row. By adding one or more else if statements: ```elif```. 

Keep in mind the ```elif``` statements that follow after a specified condition was met, are not being checked.

````python
if number > 100:
    print("number is greater than 100.")
elif number == 100:
    print("the number is 100")
elif number < 100: 
    print("number is smaller than 100")
````

If none of the statements are True, there still won't be an output. To do something for all remaining cases ```else:``` can be used.

````python
if number > 100:
    print("number is greater than 100.")
elif number == 100:
    print("the number is 100")
elif number < 100: 
    print("number is smaller than 100")
else: 
    print("I don't know what's going on")
    
````

It is also possible to check if a combination of conditions is being met. With ```or``` one can check if one or the other is it ```True``` (note: also when both conditions are ```True``` the outcome will be ```True```. So only when both are ```False``` the outcome will be ```False```.) With ```and``` one can check if both conditions are being met.

````python
weather = "cloudy"

if weather == "sunny" or weather == "rainy":
    print("wear a head")
````

The statement above can't be abbrivated to the following:

````python
if weather == "sunny" or "rainy":
    print("wear a head")
````

That is because we didn't ask python to check that the variable ```weather == "rainy"```.
We actually just asked if ```weather``` is equal to ```"sunny"``` or the string ```"rainy"``` exists.
Python assumes ```"rainy"``` exists as you just wrote it down. It being written down, proves it exists, so it is ```True```.
