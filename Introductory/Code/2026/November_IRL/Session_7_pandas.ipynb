```python
# We need to import pandas again as this is a new notebook
import pandas as pd
```

```python
# reading in our data and adding the country as the index again
data = pd.read_csv("data/gapminder_gdp_europe.csv", index_col = "country")
# The pandas documentation has other file options
# if the type this and press tab you can see all the different types of files pandas can read
# pd.read
data.head()
```

```python
# Pandas has two ways to index the data, iloc and loc
# Because there are two ways we first use a "." as it is a pandas specific form of indexing
# Numeric/positional indexing
data.iloc[0, 0] # The tables are 2D, so they work on a coordinate like system to access the values
```

```python
# taking the first three rows and the 3rd column
# We can use the slicing method like with strings and lists
data.iloc[0:3, 2]
```

```python
# this is also the first value of the dataset
# loc is used to access values by names
data.loc["Albania", "gdpPercap_1952"]
```

```python
# Using the empty ":" will then default to the first to the last
data.loc["Albania", : ] 
```

```python
# Taking all rows for 1952
data.loc[ : , "gdpPercap_1952"]
```

```python
# first three countries and the first column
data.iloc[ :3, 0]
```

```python
# Notice that these look a little different from the dataframe we loaded, let's check the type of object
type(data.iloc[ :3, 0]) # It is a pandas series, these are single columns of dataframes
```

```python
# Ranges and slicing also works with .loc and the text names of columns and rows
data.loc["Italy":"Poland", "gdpPercap_1962":"gdpPercap_1972"]
```

```python
# Checking the type of this subset of data and showing the dataset
print(type(data.iloc[ :3, 0:2]))
data.iloc[ :3, 0:2]
```

```python
# Using the results of slicing, working with our new subset to get the maximum value per column
data.loc["Italy":"Poland", "gdpPercap_1962":"gdpPercap_1972"].max()
```

```python
# Minimum value per column
data.loc["Italy":"Poland", "gdpPercap_1962":"gdpPercap_1972"].min()
```

```python
# Making a subset of the original data
subset = data.loc["Italy":"Poland", "gdpPercap_1962":"gdpPercap_1972"]
# filter for values larger than 10000
print("Where are values large?", subset > 10000)
# Because this is an assertion we get back True or False for each cell that it checks
```

```python
# Getting the index name of the max value per column in our subset (index max)
data.loc["Italy":"Poland", "gdpPercap_1962":"gdpPercap_1972"].idxmax()
```

```python
# Creating a mask to hide certain values
mask = subset > 10000
subset[mask]
```

```python
# Let's examine the stats
subset[mask].describe()
# The count actually provides the count of values, not rows
```

```python
# Checking to see if we can change the values to be rounded or integers
help(data.describe)
```

```python
# LEt's make a new mask to create a score
# This asserts that the value is greater than the mean value
mask_higher = data > data.mean()
mask_higher
```

```python
help(data.aggregate)
```

```python
# We create the filtered dataset and use aggregate to apply the sum function to the data
# The axis argument is to apply the function either row-wise or column-wise
# row-wise is axis = 0 (the default) that means it goes column-by-column and is applied to every row in the column
# column-wise is axis = 1, this means that it goes row-by-row and is applied to every column in the row
# In this case we have applied sum column-wise so each row has the sum of all it's columns created
# We then divide by the number of columns to get the ratio of years that the country had a greater than average GDP for the year
# That is the wealth score
mask_higher = data > data.mean()
wealth_score = mask_higher.aggregate("sum", axis = 1) / len(data.columns)
print(wealth_score)
```
