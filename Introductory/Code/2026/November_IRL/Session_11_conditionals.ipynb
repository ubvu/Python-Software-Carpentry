# Making Choices

It can be usefull to be able to apply an action only in a certain case. This notebook is about coding conditions. 

````python
number = 30 # We will use the variable number here as an example.

number == 30 # With "==" we can check it the value the variable number has attached to it is equal to another 
             # value (30 in this case).
             # The result of this will be boolean (either True or False).
````

It's not only possible to check if something is equal to. 
Here is a list of the other mathematical functions that can be used in a comparison:

 ```>``` : greater than
 
 ```<``` : smaller than
 
 ```==``` : equal to
 
 ```!=``` : not equal to
 
 ```>=``` : greater than or equal to
 
 ```<=``` : smaller than or equal to

```True``` and ```False``` can be used to build conditional statements, or in other words if-statements.

An if-statement has some similarity to the for loop. 
- Instead of ```for``` it only always starts with ```if```.
- Followed by the condition that ends with ```:```.
- The body of the statement is intended.

Let's explore :)

````python
number = 35
if number > 30:
    print("The number is larger than 30")
````

````python
number = 29
if number > 30:
    print("The number is larger than 30")
````

Nothing happens when the condition is ```False```. Though you could solve this by creating an other if-statement, it's more convinient to use ```else:```.

````python
number = 29
if number > 30:
    print("The number is larger than 30")

if number <= 30:
    print("The number is not larger than 30")
````

````python
number = 35
if number > 30:
    print("The number is larger than 30")
else: 
    print("The number is not larger than 30")
````

It is possible to check multiple conditions in a row. By adding one or more else if statements: ```elif```. Keep in mind the ```elif``` statements and ```else```, that follow after a specified condition was met, are not being checked.  

````python
number = 29
if number > 30:
    print("The number is larger than 30")
elif number == 30:
    print("The number is 30")
else: 
    print("The number is not larger than 30")
````

````python
# Let's checkout another example. A string can also be used in the condition.
weather = "raining" 

if weather == "raining": 
    print("wear a rain jacket")
````

It's also possible to check if an item is a part of a list. 

````python
numbers = [1,2,3,4,5,7]

1 in numbers # check if 1 is in the list of numbers
````

It is also possible to check if a combination of conditions is being met. With ```or``` one can check if one or the other is it ```True``` (note: also when both conditions are ```True``` the outcome will be ```True```. So only when both are ```False``` the outcome will be ```False```.) With ```and``` one can check if both conditions are being met. 

````python
# let's play around with the following example by adding and removing elements.

numbers = [1,2,3,4,5,7]
if (1 in numbers) and (7 in numbers):
    print("print number 1 and 7 are in the list")
elif (1 in numbers) or (7 in numbers):
     print("print either 1 or 7 is in the list")
else:
    print("neither one is in the list")
````

````python
banana = "purple"

if banana == "green" or "yellow":
    print("eat")
````

Likely, this didn't work as you expected.
That is because we didn't ask python to check that the variable ```banana == "yellow"```.
We actually just said if ```banana``` is equal to ```"green"``` or the string ```"yellow"``` exists.
Python assumes ```"yellow"``` exists as you just wrote it down. It being written down, proves it exists, so it is ```True```.

````python
# Let's checkout the adjusted version.

banana = "purple"

if banana == "green" or banana=="yellow":
    print("eat")
````
