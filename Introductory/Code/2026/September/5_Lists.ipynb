```python
# We have seen different data types stored in variables, but only single values
# it is far more useful to have multiple values stored together. One way to do this is in a list
# To store values in a list we use the [] brackets and separate the values with ","
odds = [1, 3, 5, 7]
print("the odd numbers are:", odds) # The list has the same format when printed
```

```python
# To access values in the list we "index" them. This means we use the square brackets after the name
# A number is given to indicate the position of the value we want to access
# in Python the counting of that position starts at 0, so the first value is stored in [0]
print("the first element is ", odds[0])
```

```python
print("the third element is", odds[2])
```

```python
# We test beyond the length of our list
print("the tenth element is: ", odds[9])
```

```python
# What happens when using negative numbers?
print(odds[-1])
print(odds[-3])
```

Using negative numbers means we move to the end. Since the first element is at 0 taking 1 off means you start back at the end. 

```python
# You can calculate the value inside the index too if you want
odds[1+2]
```

```python
# Indexing on strings
name = "Marcelline"
print(name[1], name[-2], name[-2])
```

```python
# We can store other data types inside of lists such as strings
names = ["Curie", "Darwing", "Turing"]
print("my favourite teacher is: ", names[2])
```

```python
# Since strings can be indexed and lists can be indexed we can first index the list to access a string
# Then we can index the string to get a specific character back
print("I don't drink ", names[2][0])
```

```python
# Fixing or changing an element in our list
names[1] = "Darwin" # correct name
print("final value of names", names)
```

```python
# Try changing the first "n" to a "w"
name = "Darnin"
```

Strings can't be changed in place (immutable), but lists can be updated in place (mutable)

```python
name[3] = "w"
print("Hi my name is slim", name)
```

```python
# risks of modifying in place (working on mutable data)

mild_salsa = ["peppers", "onion", "coriander", "tomatoes"]
hot_salsa = mild_salsa # mild salsa and hot salsa point to the same list in memory

hot_salsa[0] = "hot peppers"
print("ingredients in mild salsa:", mild_salsa)
print("ingredients in hot salsa:", hot_salsa)
```

Python has taken the two names you have given for the same list, so when you update one you update the single list. 
The list is made in the computers memory, when you add a variable name it can be referred to by name. When you set hot_salsa = mild_salsa it means that hot_salsa points to the same list in memory. So when updating the list by either name it changes the singlular list they both refer to.

To have separate lists we need to make a copy. One way we can do that for lists is calling the list() function on our first variable name.

```python
mild_salsa = ["peppers", "onion", "coriander", "tomatoes"]
hot_salsa = list(mild_salsa) # makes a copy of mild_salsa

hot_salsa[0] = "scotch bonnets" # Changing the pepper type
print("mild salsa ingredients:", mild_salsa)
print("hot salsa ingredients:", hot_salsa)
```

```python
# nested lists are lists in side of lists

veg = [["lettuce", "lettuce", "peppers", "courgette"], 
       ["lettuce", "lettuce", "peppers", "apple"], 
       ["lettuce", "coriander", "peppers", "courgette"]] # you can imagine this as the selection at a green grocers
# To access the values we now need to go two levels deep
print(veg[2][1])
```

```python
# lists can contain multiple data types at once
sample_ages = [10, 51.5, "forty"]
print(sample_ages)
```

```python
# adding values to the end of a list
odds.append(11)
print(odds)
```

```python
#removing values from a list
removed_element = odds.pop(0)
print("remaining elements:", odds)
print("removed element", removed_element)
```
