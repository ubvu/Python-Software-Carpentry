```python
# Strings can be added togther endlessly (or at least as far as your RAM will allow)
my_name = "charlie" + " " + "cameron" + " " + "greene"
my_name
```

```python
# What mathematical operations work on strings?
print(my_name * 2)
# print(my_name/2)
print(my_name - "e")
```

```python
# Sometimes you might have a mixture of data types you need to combine
number_of_pets = 5 + "2" # This doesn't work
```

```python
number_of_pets = 5 + int("2") # converting the string to an integer with the integer constructor does work
number_of_pets # Look at our new result
```

```python
# Checking the length with the inbuilt len() function
len(my_name)
```

```python
# Using the float constructor to make the calculation work
number_of_pets = 5 + float("2.2")
print(number_of_pets)
print(5 + int("2.2"))
```

```python
# We can do this directly in a print function too
print(5 + float("2"))
```
