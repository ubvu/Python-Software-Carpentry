# Lists

```python
# At the end of the last section we were introduced to lists
# Storing data is useful, but having only one value in a variable is not useful for all situations
# Lists allow us to store mulitple values in one variable
# To initiate a list we use square brackets, []
odds = [1, 3, 5, 7]
print("the odd number are: ", odds)
```

```python
# Accessing values of lists is of course vital and we can do it by 'indexing' our lists
# to index we again use the square brackets, this time after the name of the variable
# In Python indexes start at 0, that means the first element of a list is in the 0 position
print("the first element is: ", odds[0])
```

```python
# The third element is at the 2 position
print("the third element is ", odds[2])
```

```python
# What happens if we run this?
print("the tenth element is ", odds[9])
```

```python
# Why did that break?
# The list only has 4 elements and Python does ot loop over the, but instead looks for the 10th position with no luck
# What happens if we use negative numbers in the index?
# It will start at the back of the list
odds[-1]
```

```python
# Strings can also be indexed
name = "Marcelline"
print(name[1], name[-2], name[-2])
```

```python
# Since strings and lists can be indexed, how about indexing them both?
names = ["Curie", "Darwing", "Turing"]
print("my favourite letter is ", names[0][1]) # when looking to index an indexable object inside another we use a second set of []
```

```python
# Lists can be changed in place, that is to say we can update an element of a list without rewriting the whole list
names[1] = "Darwin" # correcting the name
print("final value of the names ", names)
```

```python
# Can you do the same with a string?
# Can you change the first letter of Darwin to lowercase?
name = "Darwin"
name[0] = "d"
print(name)
```

```python
# lists are mutable object and so can have a value in a single position changed or updated
# strings are immutable and so can have the whole variable updated, but not a single part
```

```python
# We can however use the 'lower' method to set the string to lower case
# Methods are functions that are attached/inherent to objects
# Strings have the lower method built into them
# they also have the 'upper' method built in
name.lower()
```

```python
name = name.lower() # updating name variable to be lowercase
name
```

There are also risks of updating mutable objects in place. Take a look at the next example where we create a list of ingredients for a recipe and use it to make another and exchange one ingredient for another.

```python
mild_salsa = ["peppers", "onions", "coriander", "tomatoes"]
hot_salsa = mild_salsa
hot_salsa[0] = "scotch bonnet" # Exchanging the peppers for the far hotter scotch bonnets
# printing both to check out our ingredient lists
print("ingredients mild: ", mild_salsa)
print("ingredients hot: ", hot_salsa)
```

Well that didn't work out. The issue is that when we created hot_salsa we actually just provided Python with a second name to use for the list. Under the hood Python has stored the list in a place in the computers memory and the name mild salsa points to that. When writing ``hot_salsa = mild_salsa`` another name began to point to it.

```python
# To fix this we can apply the list constructor to mild salsa and it will build a new list from the reference of mild_salsa
mild_salsa = ["peppers", "onions", "coriander", "tomatoes"]
hot_salsa = list(mild_salsa)
hot_salsa[0] = "ghost pepper" # A New hotter pepper because I don't like my guests
print("mild", mild_salsa)
print("hot", hot_salsa)
```

Above we used a constructor (special function) to create a new list from the reference. Constructors can be found for all data types in python and will convert the object to the new type (where possible).

```python
# This unfortunately doesn't work as they are two different types
print("I am " + 42 + " years old")
```

```python
# We also have the same possiblity for numbers
# int()
# float()
float(3) + float("14.1")
```

```python
# Lists in python can store different data types, no problemo
sample_age = [10, 51.5, "forty"]
```

```python
# The mixed data list can also be indexed without issue
sample_age[2]
```

```python
# Adding data to our list
odds.append(11) # This will add '11' to the end of the list
# We don't need to save it to the variable, the append method changes the list in place
print(odds)
```

```python
# removing data from a list with pop
print(odds.append(13)) # this updates the list, but does not return it
print(odds.pop(0)) # returns the value that we have removed
```

```python
# Pop also changes the list in place, but returns the value it has removed
removed_element = odds.pop(4)
print("remaining elements:", odds)
print("removed element:", removed_element)
```

```python
# We can also reverse the order of our lists
print(odds.reverse())
```

```python
# But it does it in place, so we need to look at the original list to see the change
print(odds)
```

```python
# The mutablity of the list persists even when using its methods
primes = odds
primes.append(2)
print(primes)
print(odds)
```

```python
# Slicing
# SLicing is the term for using a range in the indexing of an indexable object
# Slicing is done with a ':' and the number of the left is the position where your selection begins 
# and the number on the right is the first position nnot included
binomial_name = "Drosophila melanogaster"
group = binomial_name[0:10]  # from 0 up to, but no including 10
print("group", group)
```

```python
# We don't have to start at zero!
species = binomial_name[11:23]
print("species", species)
```

```python
# Another inbuilt function is `len()`, it returns the length of an object
# in this case it checks the number of characters in binomial name
len(binomial_name)
```

```python
# Using the length of the list to get the full range
print("the species is ", binomial_name[11:len(binomial_name)])
```

```python
# You can also leave part of the slice empty and python will know what to do
print("group: ", binomial_name[ :10]) # default is to start at zero
print("species: ", binomial_name[11: ]) # default is to end at the max index value
```

```python
# Something to note about the style used here
# We have kept spaces between numbers and operators, between variable names and '=', between arguments, etc
# this is all a stylistic choice that I feel makes the code easier to read and shows intention
# A clearer example of how this shows intention is with slicing
name[1:] # this could be a typo, maybe there is a missing 4 on the right hand side
name[1: ] # this looks intentional
```

```python
# One more example of slicing and a reminder on working backwards on lists
chromosomes = ["X", "Y", "2", "3", "4"]
autosomes = chromosomes[2:5]
print("autosomes:", autosomes)
last = chromosomes[-1]
print("last value: ", last)
```

```python
# Using slicing to access only the last four characters of a string or entries of a list
string_for_slicing = 'Observation date: 02-Feb-2013'
list_for_slicing = [['fluorine', 'F'],
                    ['chlorine', 'Cl'],
                    ['bromine', 'Br'],
                    ['iodine', 'I'],
                    ['astatine', 'At']]

# You can use a third value in your slicing to set the step size. This detemines how many elements the section moves
# along by. The default size is one. 
test_list = [1,2,3,4,5,6,6,76,87,8,10]
print(test_list[1:9:3])
# Can you use the step size argument to create a new string that contains only every other character and store it in a new variable
beatles = "In an octopus's garden in the shade"
```

```python
# Starting at -4 takes us to the fourth last character, leaving the second half empty will take it to the end of the string
sliced_string = string_for_slicing[-4: ]
print(sliced_string)
```

```python
# Exactly the same appproach works here! :)
sliced_list = list_for_slicing[-4: ]
print(sliced_list)
```

```python
print(beatles[ : :2]) # As we saw with test list the third value is the step
# This lets the default value fill in for the start and end of the list and the third value makes it skip every other character
```
