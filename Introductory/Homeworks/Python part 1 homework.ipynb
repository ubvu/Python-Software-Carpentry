<div class=\"alert alert-success\">

    import pandas as pd
    # First we read in out data set and assign it to a variable
    df = pd.read_csv("Dushane_research_data.csv")
    # when we put 'df' as the last line of a cell we can look at a formatted version of the dataframe
    df

    # It turns out that there are no column names in Dushane's data
    # We can deal with that when reading in the data by using the "header" argument
    df = pd.read_csv("Dushane_research_data.csv", header=None)
    # We want to look at the maximum value per week (column)
    weekly_max = df.max(axis=0) # axis 0 means that the method is applied 'column-wise'
    # We can look through this list to find the largest, or we can code it
    weekly_max = weekly_max.sort_values(ascending=False)
    print(weekly_max) # the order is set to NOT be ascending by using False
    # We can now see that the column "5" has the most growth
    # But remember, the column names started on 0, so week 6 had the most growth

    # We can repeat this with the min method on our dataframe
    # This time we set the ascending argument to True
    weekly_min = df.min(axis=0).sort_values(ascending=True)
    print(weekly_min)
    # Week 15 had the slowest growth (remember week labels started on 0)
