    def team_averages(dataframe, laps, team_col):
        # We need the dataframe, the list of laps column names, and the team column name
        dataframe["swimmer_avg"] = dataframe[laps].mean(axis=1) # Getting the mena per swimmer
        # Repeating the team avg process
        team_avgs = dataframe[[team_col, "swimmer_avg"]].groupby(team_col).mean()
        # Don't forget to return your output, without this you won't see the results!
