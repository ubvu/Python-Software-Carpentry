### .sav (SPSS)

To work with the SPSS data file type .sav we will use a package called [pyreadstat](https://github.com/Roche/pyreadstat). It is available as a package on pypi, meaning we can simply [pip install](https://pip.pypa.io/en/stable/cli/pip_install/) it.

There are two parts to a sav file, the data and the labels. pyreadstat refers to the labels (and additional data) as the metadata, so when opening an SPSS data file there are two variables needed to store it.

```python
import pyreadstat

df, meta = pyreadstat.read_sav("job_satisfaction.sav")
# Take a look at the data, what format is it?
df
```

The name given to the variable probably gave it away, but this packages can open the data from sav file directly into a pandas dataframe. Super useful!

```python
# What about the metadata?
meta
```

The metadata_container is exactly what it says, it is a holder for all the information in the data that is not explicitly the numerical data itself. If you scroll along you will see some very useful things, such as the variable_value_labels because Gender = 0 or Gender = 1 is useful for calculations, but not for our human insight. There is also a column_label list because columns are not always named in a way that outsiders can easily decipher. Both of these are important contextual clue about the dataset as a whole.

If you open up the file we just saved with excel, libre office or another open source tool you'll notice an additional column in the data, the index. We can leave out the index column with another argument "index=False" as the default is True. 


One of the useful things about xlsx files is that they can have more than one "sheet". When reading in an excel file we can choose the page we want to read, but updating a sheet is a bit more difficult. You will notice that our output file has only one sheet and it has not retained the name. The default saving of excel files in pandas only works with a single sheet file, let's take a look at how to update existing files to allow for multiple sheets files.

First we create an ExcelWriter object that we feed to the ".to_excel()" method so it knows how we want it to write out the data. "pd.ExcelWriter()" has a few arguments that are important here, the first is the file we are targeting, the "engine" argument we will tell in this case to use "openpyxl" because it has the features we are after. In this example we are appending to our initial output so we set the "mode" as "a", for append. The default mode is "w", for write. The last argument that is interesting for us is the "if_sheet_exists" argument that decides what to do if the sheet you name already exists in the file. In this situation we are updating the dataframe from that sheet and so will "replace" it entirely. 

The task here is to open the second sheet in "Temperatures.xlsx" file, filter to keep only the cities with less than 23 temperature and append a sheet called "City" to the output file from the previous step without an index column?


Can you create a variable 'writer' to contain the pd.ExcelWriter object

```python
writer = pd.ExcelWriter(out_file, engine='openpyxl', mode='a', if_sheet_exists="replace")
print("writer made")
sheet_1_clean.to_excel(writer, sheet_name=sheet_1,index=False)
sheet_2_clean.to_excel(writer, sheet_name=sheet_2,index=False)
print("pandas complete")
writer.close()
```
