# Repeating actions with for loops

Copy-pasting can be usefull for repeating actions. However it's: 
1. not scalable
2. high maintenance
3. fragile

Let's explore these downsides of copy pasting first. Before we dive into one of the alternatives, the for loop.


````python
# Let's imagine we have the list odds. With items that we would like to print one by one.
odds = [1, 3, 5] 
print(odds[0])
print(odds[1])
print(odds[2])
````

````python
# Now the list gets longer 
odds = [1, 3, 5, 7] 
# we can copy the previous code and adjust
print(odds[0])
print(odds[1])
print(odds[2])
print(odds[3]) # we need one print statement, otherwise the last item won't be printed 
````

````python
# Now the list gets one shorter 
odds = [1, 3, 5] 
# we can copy the previous code for printing - however this will result in an error
print(odds[0])
print(odds[1])
print(odds[2])
print(odds[3]) # This element no longer exists in the list
````

````python
# Though on this small scale this is easy to solve, when having larger bits of code than a printing statement or longer lists,
# this method becomes less suitable.
````

### For loop

A for loop is a great way to automate repeating an action. This ranges from applying the same analysis on a ranges of different files, to building a simple model that simulates the amount of water in a bathtub. However large Earth System Models that simulate the climate, are also build-up of many of these for loops.

The for loop has a specific structure. The first line defines what you like to loop over (the collection, that can be a list like odds). The variable is one after the next each item from the collection one loops over. The loop goes one-by-one through this collection, until there are no items left, and excecute what ever is in the body of the loop. The body of the loop is intended. The body of the loop ends where the indentation ends.  

    for variable in collection:
        # do things using variable, such as print



It always starts with the ```for```, followed by the variable (you can give this any name), followed always by ```in``` (this is a part of the for loop format), followed by the collection you would like to loop over (e.g. the list odds), ending with ```:```. 

````python
for item in odds[0:3]:
    print(item)
    
````

````python
for item in odds[0:3]:
print(item) 
    
````

````python
# The body of the for loop needs to be indented. 
# One needs to add a least one line of code to the for loop. Otherwise it will give an error.
# However adding multiple lines is possible
````

````python
for item in odds[0:3]:
    print(item)
    print(item)
    print(item)
    print(item)
print("the end")
````

````python
# One can build a counter with a for loop. Let's try this out, by writing a piece of code that computes the number of names a list, making use of the 
# for loop.

names = ["Curie", "Darwin", "Turing"] # define the is of names
length = 0 # create a variable that we can use to count with, and setting it to zero.

for name in names: # start the for loop
    length = length + 1 # add 1 each iteration to the variable lenght, our variable that keeps track of the number of iterations
    print(length) # The variable length

print("There are", length, "names in the list.")
````
