```python
# We start by importing pandas again
import pandas as pd
# We want this dataset to use th country name as the index for clarity during our analysis
data = pd.read_csv("data/gapminder_gdp_europe.csv", index_col = "country")
# We can index the data by value with .iloc[]
# It is indexing so [] are needed and it is 2-dimensional so we need two values. 
# Think of it like co-ordinates of the table starting with 0,0 in the top left cell
# The first value is the row and the second value is the column
print(data.iloc[0,0])
```

```python
# Checking the first few rows to confirm that the .iloc[0,0] is correct
data.head()
```

```python
# We can refer to the row by name using .loc
# The first value is the row, the second value is the column
print(data.loc["Albania", :]) # Using the ":" without values means that we let python default to getting all values
```

```python
# Referring to column by name when using loc
print(data.loc[:, "gdpPercap_1952"])
```

```python
# slicing our data with ":"
print(data.loc["Italy":"Poland", "gdpPercap_1962":"gdpPercap_1972"])
```

```python
# the toughest bugs to fix are those without errors
# Making a typo here it seems to mean that pandas will ignore the typo and instead takes it as a blank value meaning 
# it reads "10:t19" as "10:" instead of "10:t19" and an error
print(data.loc["Italy":"Poland", "gdpPercap_1962":"gpdPercap_1972"].max())
```

```python
# operating on our sliced data in place
# finding the max value for the countries per year
print(data.loc["Italy":"Poland", "gdpPercap_1962":"gdpPercap_1972"].max())
```

```python
# and the minimum
print(data.loc["Italy":"Poland", "gdpPercap_1962":"gdpPercap_1972"].min())
```

```python
# We can of course save our subset to a variable and then use it
subset = data.loc["Italy":"Poland", "gdpPercap_1962": "gdpPercap_1972"]
print("subset of data: \n", subset) # "\n" is a special character in python, it means new line
```

```python
# Which values are greater than 10k?
print("where are values large? \n", subset > 10000)
```

```python
# here we make a mask for our dataframe
mask = subset > 10000
print(subset[mask]) # applying the mask removes/hides values that get masked
```

```python
# when looking at the describe check out the counts
# The counts actually tell us how many values there are in the column, not how many rows
# NaN is empty, as in no value of any kind
print(subset[mask].describe()) 
```

```python
# We can combine the masking with a calculation
# This will show you the fraction of the time that a country has had greater than continental average gdp
# You can always print out each part of this to see what it looks like every stage along the way
# First is the mask, next is the aggregation with a "sum" method run on axis = 1
# The axis refer to whether we want to run row-wise or column-wise
# axis=0 is row-wise meaning it takes one row and runs across it column-by-column
# axis=1 is column-wise and will take one column and run down it row-by-row
# A little confusing, but you could imagine that the 1 is vertical, so it will run vertically
mask_higher = data > data.mean()
wealth_score = mask_higher.aggregate('sum', axis=1) / len(data.columns)
print(wealth_score)
```

```python
# Lastly we will take a look at grouping our data with the groupby method
# We ask python to sum the values of the groups
# This means each country with the same wealth score is grouped together per year and their total gdp is shown
print(data.groupby(wealth_score).sum())
```

```python
# without using a method on the groupby method we end up with just a referral to the method rather than a result
data.groupby(wealth_score)
```
