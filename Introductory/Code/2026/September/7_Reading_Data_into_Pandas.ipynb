In this lesson part we use the library pandas to open tabular data (tables, think excel) and clean and analyse it

```python
# first we need to import pandas to be able to use it
# We include the alias pd by using import  as 
import pandas as pd
# We read the data with the function read_csv
# Inside this we put the path to the dataset
# Despite it being read_csv we still need to include the file extension (".csv")
data_oceania = pd.read_csv("data/gapminder_gdp_oceania.csv")
data_oceania
```

```python
# Explicit path to the file
pd.read_csv("C:/Users/tis254/data/gapminder_gdp_oceania.csv")
```

```python
# We can get some insights into the dataset using the "info" method
# It is a method (function that is inherent to the object) so we access it with a "." and use "()" after it
data_oceania.info()
```

```python
# DataFrames also have attributes (fixed values associated with them)
# One example is the columns
print(data_oceania.columns)
```

```python
# If we want, we can even transpose our dataset
# This will flip it so that columns are rows and rows are columns
print(data_oceania.T)
```

```python
# by not printing we allow the notebook to format our table nicely
data_oceania.T
```

```python
# There is another way to get a summary of the dataframe
# describe provides some basic stats on the dataframe per column
# std is the standard deviation
print(data_oceania.describe())
```

```python
# This returns the fact that it is a method on said table
# It happens because we haven't included the () after the function name
data_oceania.describe
```

```python
# We can always get information on functions and methods using the help() function of a "?"
# The describe method does not need an argument (entry in the function), but it can use percentiles, include, or exclude to filter
?data_oceania.describe
```

```python
# Can you read in the americas dataset into the variable data_americas and display the summary stats?
data_americas = pd.read_csv("data/gapminder_gdp_americas.csv")
print(data_americas.describe())
```

```python
# After reading the data use help(data_americas.head) and help (data_americas.tail) to find out about the head and tail methods
print(data_americas.head())
print(data_americas.tail())
```

```python
# Can you display the last three columns of data by making use of another method we tested above?
data_americas.T.tail(3)
```

```python
# We can include another argument in the read_csv function called index_col
# This will take the column "country" and set it as the table index
# Look above and you'll see that data_americas.tail() for instance has numbers along the left hand side
# These numbers are the index values, you can refer to the row by this number
data_americas_2 = pd.read_csv("data/gapminder_gdp_americas.csv", index_col="country")
data_americas_2.head()
```

```python
# Now when we transpose the table we have the country name as the column name
# It is far clearer than the index numbers
data_americas_2.T.tail(3)
```

```python
# pandas does of course allow you to save your data as well
# to save a file as a csv we use the method (a function that is inherently attached to an object) ".to_csv()"
# The only argument (value) that needs to be given to the method is the location of where we want to save the file and its name and file type
# despite it being to_csv we still need to explicitly state that it is a csv file
data_americas_2.to_csv("data/gapminder_gdp_americas_2.csv")
```
