# Reading and working with data 

In order to work with our data we will use another common library, pandas. It has many, many useful features and so you can always find out more in [its documentation pages](https://pandas.pydata.org/docs/). There is even a version of pandas for geographical data, [geopandas](https://geopandas.org/en/stable/)! 

```python
# Importing pandas with its standard alias
# It is already a short name, but this is the industry standard
import pandas as pd
```

```python
# If the importing doesn't work because Python can't 'find the file or directory' it could be a typo (don't worry, these happen all the time)
# It could also be that your notebook isn't running from where you think it is
# We can check that with the lines of code below
import os
print(os.getcwd())
```

```python
# We want to use our data, so we read it into a variable
# To do this we can use the pandas read_csv function
# There is only one argument that we need to fill out, that is the explicit or relative file path (including the file type at the end!)
data_oceania = pd.read_csv("data/gapminder_gdp_oceania.csv")
print(data_oceania) # Looking at our dataset
```

### Pandas can read in tabular (table like) data and stores it in its own data format, a dataframe. 

```python
# One of the benefits of using a notebook is that we can get nicely formatted datasets if we dispay it this way
data_oceania
```

This dataset is the GDP through the years  of two of the countries in the continent Oceania. The bold names are the column names and on the far left you can see 0 and 1, these are the index values.

```python
# Display first 5 rows in structured way
data_oceania.head()
```

```python
# display the last 5 rows in a structured way
data_oceania.tail()
```

```python
# Reloading the data where the index is the country column instead of numeric
data_oceania_country = pd.read_csv("data/gapminder_gdp_oceania.csv", index_col = "country")
data_oceania_country
```

```python
# There are a lot of arguments that the read_csv() function can use, to find out more you can call the help() function on it
help(pd.read_csv)
# In fact the help function works on any function
```

```python
# Get some information (metadata) about our dataframe
data_oceania_country.info()
# We have the index value of the columns, the number of not empty values in the column, and the datatype of the column
# float64 is a 64bit float
```

```python
# We can print out the columns by accessing the attribute of dataframe
# An attribute is an inbuilt variable attached to the data object, much like a method is an inherent function for a data object
print(data_oceania_country.columns)
```

```python
# If we want to use the columns and index them we can convert the object to a list!
list(data_oceania_country.columns)
```

```python
# The describe function provides us with some simple statistics for each column
# The std is standard deviation and the percentages of the lower, mid, and upper quartiles
data_oceania_country.describe()
```
