```python
# trying a more complex problem

new_list = ["first element", 2, "third element", 4, "five"]

for value in new_list:
    if value < 5:
        print("very small value, what a shame")
    elif value == "five":
        print("What a beautiful value")
    else:
        print(value * 10, "don't care")
# this breaks because python does not know what the numerical value of "first element is"
# It would break just the same if you put your nephew's name in because python does not have the
# context to know that "ian" is less than 5 years old
```

```python
# to get around the error we add a number as the first element
new_list = [0, "first element", 2, "third element", 4, "five"]

for value in new_list:
    if value < 5:
        print("very small value, what a shame")
    elif value == "five":
        print("What a beautiful value")
    else:
        print(value * 10, "don't care")

# Now it runs once and then breaks, the 0 can be evaluated against the < 5 successfully
```

```python
# We might want more information though
# imagine you are running an analysis on a dataset with 10,000 entries and somewhere partway through it breaks

# that is where printing comes in handy

new_list = [0, "first element", 2, "third element", 4, "five"]

for value in new_list:
    print("Value is:", value)
    if value < 5:
        print("very small value, what a shame")
    elif value == "five":
        print("What a beautiful value")
    else:
        print(value * 10, "don't care")

# Now we see which value runs and which it breaks on
# that extra context will help us to fix our if else statements or clean our data
```
